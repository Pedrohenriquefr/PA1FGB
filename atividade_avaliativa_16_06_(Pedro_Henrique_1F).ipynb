```python
#QUESTÂO 02 - Soma dos Números (2 pontos)

#Solicite ao usúario um número inteiro positivo
#Utilizando um laço de repetição, calcule e exiba a soma de
#todos os números de 1 até o numero informado

#Exemplo:

#Digite um número: 5
#Resultado: 15

#Pois:

#1 + 2 + 3 + 4 + 5 = 15

número = int(input("Digite um número:"))
soma = 0
for i in range(1, número + 1):
    soma += i
print("A soma dos números de 1 a", número, "é", soma)
```

```python
#QUESTÃO 04 – Fatorial (4 pontos)
#Solicite ao usuário um número inteiro positivo e calcule
#seu fatorial utilizando um laço de repetição.

#Exemplo:

#Digite um número: 5

#Fatorial = 120

#Pois:

#5! = 5 × 4 × 3 × 2 × 1

##O programa deve validar se o número informado é positivo.
numero = int(input("Digite um número:"))
fatorial = 1
for i in range(1, numero + 1):
    fatorial *= i
if numero < 0:
    print("O número deve ser positivo.")
else:
  print("O fatorial de", numero, "é", fatorial)


```

```python
#QUESTÃO 05 – Caixa Eletrônico (5 pontos)


#Um caixa eletrônico possui notas de:

#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1

#Solicite ao usuário o valor do saque e informe a menor
#quantidade possível de notas para realizar o pagamento.

#Exemplo:

#Digite o valor do saque: 287

#Saída esperada:

#2 nota(s) de R$100
#1 nota(s) de R$50
#1 nota(s) de R$20
#1 nota(s) de R$10
#1 nota(s) de R$5
#1 nota(s) de R$2

#Requisitos:
#- Utilizar laços de repetição.
#- Não utilizar listas.
#- Exibir apenas as notas utilizadas.

saque = int(input("Digite o valor do saque: "))

nota = 100

while nota >= 2:
    qtd = saque // nota

    if qtd > 0:
        print(f"{qtd} nota(s) de R${nota}")

    saque %= nota

    if nota == 100:
        nota = 50
    elif nota == 50:
        nota = 20
    elif nota == 20:
        nota = 10
    elif nota == 10:
        nota = 5
    elif nota == 5:
        nota = 2
    else:
      break

```
