```python
#1 - Par ou ímpar
#Peça um número ao usuário e informe se ele é par ou ímpar.
numero = int(input("Digite um número: "))
if numero % 2 == 0:
    print("O número é par.")
else:
    print("O número é ímpar.")
```

```python
#2 - Maior de dois números
#Solicite dois números e mostre qual é o maior (ou se são iguais).
numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input("Digite o segundo número: "))
if numero1 > numero2:
    print("O primeiro número é maior.")
elif numero2 > numero1:
    print("O segundo número é maior.")
else:
    print("Os números são iguais.")
```

```python
#3 - Aprovação de aluno
#Peça a nota de um aluno e informe:
#Aprovado (nota => 7)
#Recuperação (5 <= nota < 7)
#Reprovado (nota < 5)
nota = float(input("Digite a nota do aluno:"))
if nota >= 7:
    print("Aprovado")
elif nota >= 5:
    print("Recuperação")
else:
    print("Reprovado")

```

```python
#4 - Classificação de idade
#Receba a idade e classifique:
#Criança (0–12)
#Adolescente (13–17)
#Adulto (18–59)
#Idoso (60+)
idade = int(input("Digite a idade:"))
if idade >= 0 and idade <= 12:
    print("Criança")
elif idade >= 13 and idade <= 17:
    print("Adolescente")
elif idade >= 18 and idade <= 59:
    print("Adulto")
elif idade >=60:
    print("Idoso")
```

```python
#5 - Senha correta
# Defina uma senha fixa no código. Peça ao usuário para digitar e informe se
# está correta ou incorreta
senha_correta = "140426"
senha_digitada = input("Digite a senha:")
if senha_digitada == senha_correta:
    print("Acesso Concedido!")
else:
    print("Acesso Negado!")
```

```python
#6 - Positivo, negativo ou zero
#Peça um número e diga se ele é positivo, negativo ou zero.
numero = int(input("Digite um número:"))
if numero > 0:
    print("O número inserido é: Positivo!")
elif numero < 0:
    print("O número inserido é: Negativo!")
else:
    print("O número inserido é: Zero!")
```

```python
#7 - Calculadora simples
#Peça dois números e uma operação (+, -, *, /). Use if/elif para calcular o resultado.
numero1 = int(input("Digite o primeiro número:"))
numero2 = int(input("Digite o segundo número:"))
operacao = input("Digite a operação (+, -, *, /):")
if operacao == "+":
    print(numero1 + numero2)
    print("A soma dos números é:", numero1 + numero2)
elif operacao == "-":
    print(numero1 - numero2)
    print("A subtração dos números é:", numero1 - numero2)
elif operacao == "*":
    print(numero1 * numero2)
    print("A multiplicação dos números é:", numero1 * numero2)
elif operacao == "/":
    print(numero1 / numero2)
    print("A divisão dos números é:", numero1 / numero2)
else:
    print("Operação inválida!")
```

```python
#8 - Dia da semana
#Peça um número de 1 a 7 e mostre o dia correspondente (ex: 1 = domingo).
numero = int(input("Digite um número de 1 a 7:"))
if numero == 1:
    print("Domingo")
elif numero == 2:
    print("Segunda-feira")
elif numero == 3:
    print("Terça-feira")
elif numero == 4:
    print("Quarta-feira")
elif numero == 5:
    print("Quinta-feira")
elif numero == 6:
    print("Sexta-feira")
elif numero == 7:
    print("Sábado")
else:
    print("Número inválido!")
```

```python
#9 - Ano bissexto
#Solicite um ano e informe se ele é bissexto ou não.
ano = int(input("Digite um ano:"))
if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print("O ano é bissexto!")
    print(ano)
else:
    print("O ano não é bissexto!")
    print(ano)
```

```python
#10 - Desconto em compras
#Peça o valor de uma compra:
#≥ 100: 10% de desconto
#≥ 50 e < 100: 5% de desconto
#< 50: sem desconto
valor_compra = float(input("Digite o valor da compra:"))
if valor_compra >= 100:
    valor_com_desconto = valor_compra * 0.9
    print("O valor com desconto é:", valor_com_desconto)
elif valor_compra >= 50:
    valor_com_desconto = valor_compra * 0.95
    print("O valor com desconto é:", valor_com_desconto)
else:
    print("O valor com desconto é:", valor_compra)


```
